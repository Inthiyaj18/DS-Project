```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.plot(x, y, color='green', linestyle='--', marker='o',ms='10')
plt.title("Line Chart Example")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.grid(linestyle='--',alpha=0.3,color='r')
plt.show()

```

```python
import matplotlib.pyplot as plt
categories = ['Apples', 'Bananas', 'Oranges']
sales = [150, 200, 120]
colors = ['green', 'yellow', 'orange']
plt.bar(categories, sales, color=colors, width=0.6, edgecolor='black', label='Fruit Sales')
plt.xlabel('Fruit Type')
plt.ylabel('Units Sold')
plt.title('Quarterly Fruit Sales')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
sizes = [15, 30, 45, 10]
labels = ['Frogs', 'Hogs', 'Dogs', 'Logs']
plt.pie(sizes, labels=labels)
plt.show()
```

```python
labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 30, 20, 10]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Language Market Share")
plt.show()

```

```python
x = [1, 2, 3, 4, 5]
y1 = [5, 10, 15, 20, 25]
y2 = [25, 20, 15, 10, 5]

plt.subplot(1, 2, 1)
plt.plot(x, y1, color='green')
plt.title("Increasing")

plt.subplot(1, 2, 2)  
plt.plot(x, y2, color='red')
plt.title("Decreasing")

plt.show()




```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.plot(x, y, color='b', linestyle='-', marker='d',ms='10',mfc='y',mec='grey')
plt.title("Line Chart Inthiyaj")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.grid(linestyle='--',alpha=0.3,color='r')
plt.show()

```
